```python

import pandas as pd
import numpy as np
```

```python

df = pd.read_csv("viratkohli.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python

print("Dataset Information:")
df.info()

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nStatistical Summary:")
display(df.describe())
```

```python

print("Missing values in each column:")
print(df.isnull().sum())
```

```python

# Keep an unchanged copy of the original data
cleaned_df = df.copy()

# Remove the unnecessary index column
if "index" in cleaned_df.columns:
    cleaned_df = cleaned_df.drop(columns=["index"])

# Remove extra spaces from text columns
text_columns = ["opponent", "ground", "match"]

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].str.strip()

# Convert date from text to datetime
cleaned_df["date"] = pd.to_datetime(
    cleaned_df["date"],
    format="%d%b%Y",
    errors="coerce"
)

# Check for invalid dates
print("Invalid dates:", cleaned_df["date"].isna().sum())

# Remove exact duplicate records if any
cleaned_df = cleaned_df.drop_duplicates()

print("Cleaning completed!")
display(cleaned_df.head())
```

```python

print("Cleaned Dataset Shape:")
print(cleaned_df.shape)

print("\nMissing Values:")
print(cleaned_df.isnull().sum())

print("\nData Types:")
print(cleaned_df.dtypes)

print("\nDuplicate Records:")
print(cleaned_df.duplicated().sum())
```

```python

cleaned_df.to_csv(
    "cleaned_viratkohli.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
